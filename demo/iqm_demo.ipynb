# IQM demo

### Solver usage

```python
from QHyper.problems.knapsack import KnapsackProblem
problem = KnapsackProblem(max_weight=2,
                          item_weights=[1, 1, 1],
                          item_values=[2, 2, 1])
```

```python
from QHyper.solvers.gate_based.iqm import QAOA
from QHyper.optimizers import OptimizationParameter
from QHyper.optimizers.dummy import Dummy

solver = QAOA(problem,
    layers=5,
    gamma=OptimizationParameter(init=[0.25, 0.25, 0.25, 0.25, 0.25]),
    beta=OptimizationParameter(init=[-0.5, -0.5, -0.5, -0.5, -0.5]),
    optimizer=Dummy(),
    penalty_weights=[1, 2.5, 2.5],
    backend_url="https://cocos.resonance.meetiqm.com/garnet",
    backend_token="<YOUR_BACKEND_TOKEN>",
    shots=1000
)
```

```python
solver.solve()
```

### Hyper optimizer usage

```python
from QHyper.optimizers import OptimizationParameter
from QHyper.solvers.hyper_optimizer import HyperOptimizer
from QHyper.optimizers.grid_search import GridSearch

solver = QAOA(problem,
    layers=2,
    gamma=OptimizationParameter(min=[0.0, 0.0],
                                init=[0.5, 0.5],
                                max=[6.28, 6.28]),
    beta=OptimizationParameter(min=[0.0, 0.0],
                            init=[1.0, 1.0],
                            max=[6.28, 6.28]),
    optimizer=Dummy(),
    backend_url="https://cocos.resonance.meetiqm.com/garnet",
    backend_token="<YOUR_BACKEND_TOKEN>",
    shots=1000
)

hyper_optimizer = HyperOptimizer(
    optimizer=GridSearch(verbose=True, disable_tqdm=False),
    solver=solver,
    penalty_weights={
        "min": [1, 1, 1],
        "max": [2.1, 2.1, 2.1],
        "init": [1, 1, 1],
        "step": [1, 1, 1]
    }
)
```

```python
hyper_optimizer.solve()
```

```python
hyper_optimizer.run_with_best_params()
```
